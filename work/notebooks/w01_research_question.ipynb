## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = len(df)
down_trend_df = df[df["trend_direction"] == "down"]
percent_down_trend = (len(down_trend_df) / total_pages) * 100
median_impressions_down = down_trend_df["impressions_90d"].median()

print(f"Number of pages analyzed: {total_pages}")
print(f"Pages showing an observed downward trend: {len(down_trend_df)} ({percent_down_trend:.1f}% of total)")
print(f"Median impressions of downward-trending pages: {median_impressions_down:.0f}")

print("\nTop 5 down-trending pages with high impressions (high impact candidates for content refresh):")
display(down_trend_df.sort_values(by="impressions_90d", ascending=False).head(5))
```

### 4. Careful words: what I can and can't claim

* **What I can claim:**

* **What I cannot claim:**
